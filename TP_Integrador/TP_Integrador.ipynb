```python
import os
import time
from huggingface_hub import InferenceClient

cliente = InferenceClient(
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    token=os.getenv("HF_TOKEN")
)

inicio = time.time()

try:
    resultado_remoto = cliente.text_classification(
        "This movie was absolutely amazing!"
    )

    fin = time.time()

    print("Resultado:", resultado_remoto)
    print("Tiempo:", round(fin - inicio, 4), "segundos")
    print("Tipo de resultado:", type(resultado_remoto))

except Exception as e:
    print("Error:", type(e).__name__)
    print("Detalle:", str(e))
```

**Error 402 Payment Required — InferenceClient**

Al probar la inferencia remota con Hugging Face, inicialmente se produjo un error 403 Forbidden porque el token no tenía los permisos necesarios para utilizar Inference Providers. Se corrigieron los permisos y la autenticación pasó a funcionar.

Posteriormente, la solicitud devolvió un error 402 Payment Required, indicando que la cuenta no tenía créditos disponibles para realizar inferencias remotas. Por este motivo, no fue posible medir el tiempo de respuesta del servicio remoto.

La alternativa local funcionó correctamente y permitió comparar ambos modelos con las mismas ocho frases: los dos obtuvieron un 75 % de aciertos. DistilBERT tardó 0,3697 segundos y XLM-RoBERTa, 0,6581 segundos en procesar las ocho frases. Estas mediciones corresponden a ejecuciones locales con los modelos ya cargados.


### Conclusiones de la comparación de modelos

Probamos dos modelos de análisis de sentimientos con las mismas ocho frases. DistilBERT obtuvo 6 aciertos de 8 (75 %) y XLM-RoBERTa también obtuvo 6 aciertos de 8 (75 %).

En las mediciones locales, DistilBERT tardó 0,3697 segundos en procesar las ocho frases, con un promedio de 0,0462 segundos por frase. XLM-RoBERTa tardó 0,6581 segundos, con un promedio de 0,0823 segundos por frase.

En esta prueba, DistilBERT fue más rápido, aunque ambos obtuvieron el mismo porcentaje de aciertos. Los dos modelos tuvieron dificultades con expresiones informales o ambiguas, como «This movie was sick!». XLM-RoBERTa clasificó «Meh.» como neutral, mientras que DistilBERT la clasificó como negativa.

Estos resultados corresponden a un conjunto pequeño de ocho frases y no permiten afirmar que los modelos tengan una precisión general del 75 %.

También probamos InferenceClient para ejecutar el modelo en un servidor remoto de Hugging Face. La solicitud terminó con un error HTTP 402 porque no había créditos disponibles. Por ese motivo, no pudimos medir el tiempo de respuesta remoto ni compararlo directamente con la ejecución local.

### Análisis de InferenceClient

`InferenceClient` permite enviar una solicitud a un proveedor remoto de inferencia de Hugging Face. A diferencia de `pipeline`, que ejecuta el modelo en el entorno local, esta alternativa necesita conectarse a un servicio externo.

En nuestro experimento, configuramos un cliente con el modelo `distilbert/distilbert-base-uncased-finetuned-sst-2-english` y utilizamos el método `text_classification()` para enviar una frase.

La solicitud no devolvió una predicción porque el servicio respondió con el error HTTP 402 Payment Required, indicando que no había créditos disponibles. Por lo tanto, no pudimos comprobar el tipo de retorno de una respuesta remota exitosa ni medir su tiempo de ejecución.

Este resultado también es parte del experimento: registramos el error real y evitamos inventar una respuesta o un tiempo de ejecución que no obtuvimos.
